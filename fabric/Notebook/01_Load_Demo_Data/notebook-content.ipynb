```python
from pyspark.sql.functions import *
base_path = 'Files/demo-data'

# dim_customer
df = spark.read.option('header','true').option('inferSchema','true').csv(f'{base_path}/dim_customer.csv')
df = df.withColumn('opted_in', col('opted_in').cast('boolean')).withColumn('last_visit', to_date(col('last_visit')))
df.write.mode('overwrite').format('delta').saveAsTable('dim_customer')
print(f'dim_customer: {df.count()} rows')

# dim_campaign
df = spark.read.option('header','true').option('inferSchema','true').csv(f'{base_path}/dim_campaign.csv')
df = df.withColumn('start_date', to_date(col('start_date'))).withColumn('end_date', to_date(col('end_date')))
df.write.mode('overwrite').format('delta').saveAsTable('dim_campaign')
print(f'dim_campaign: {df.count()} rows')

# dim_store
df = spark.read.option('header','true').option('inferSchema','true').csv(f'{base_path}/dim_store.csv')
df.write.mode('overwrite').format('delta').saveAsTable('dim_store')
print(f'dim_store: {df.count()} rows')

# dim_product
df = spark.read.option('header','true').option('inferSchema','true').csv(f'{base_path}/dim_product.csv')
df = df.withColumn('is_seasonal', col('is_seasonal').cast('boolean'))
df.write.mode('overwrite').format('delta').saveAsTable('dim_product')
print(f'dim_product: {df.count()} rows')

# dim_weather
df = spark.read.option('header','true').option('inferSchema','true').csv(f'{base_path}/dim_weather.csv')
df = df.withColumn('date', to_date(col('date')))
df.write.mode('overwrite').format('delta').saveAsTable('dim_weather')
print(f'dim_weather: {df.count()} rows')

# fact_transaction
df = spark.read.option('header','true').option('inferSchema','true').csv(f'{base_path}/fact_transaction.csv')
df = df.withColumn('timestamp', to_timestamp(col('timestamp')))
df.write.mode('overwrite').format('delta').saveAsTable('fact_transaction')
print(f'fact_transaction: {df.count()} rows')

# fact_impression
df = spark.read.option('header','true').option('inferSchema','true').csv(f'{base_path}/fact_impression.csv')
df = df.withColumn('clicked', col('clicked').cast('boolean')).withColumn('converted', col('converted').cast('boolean')).withColumn('timestamp', to_timestamp(col('timestamp')))
df.write.mode('overwrite').format('delta').saveAsTable('fact_impression')
print(f'fact_impression: {df.count()} rows')

# identity_match
df = spark.read.option('header','true').option('inferSchema','true').csv(f'{base_path}/identity_match.csv')
df.write.mode('overwrite').format('delta').saveAsTable('identity_match')
print(f'identity_match: {df.count()} rows')

print('All tables loaded!')
```
